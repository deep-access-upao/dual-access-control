# Comparación de mejora del modelo siamés
Carga los resultados pequeños ya generados; no calibra thresholds ni ejecuta entrenamiento.

```python
import json
from pathlib import Path
import matplotlib.pyplot as plt
import pandas as pd

root = Path('..') if Path.cwd().name == 'notebooks' else Path('.')
results = root / 'outputs' / 'experiments' / 'model_comparison'
thresholds = pd.read_csv(results / 'threshold_metrics.csv')
stress = pd.read_csv(results / 'stress_metrics_security_first.csv')
summary = json.loads((results / 'comparison_summary.json').read_text(encoding='utf-8'))
```

## Validation y test limpio

```python
columns = ['model', 'criterion', 'threshold', 'validation_f1', 'validation_far', 'validation_frr', 'test_f1', 'test_far', 'test_frr']
thresholds[columns].round(4)
```

## FRR fotométrico con security_first

```python
conditions = ['low_light', 'overexposure', 'low_contrast']
photo = stress[stress['condition'].isin(conditions)]
table = photo.pivot(index='condition', columns='model', values='frr').reindex(conditions)
display(table.round(3))
ax = table.plot.bar(figsize=(8, 4), ylim=(0, 1), title='FRR bajo degradación fotométrica')
ax.set_ylabel('FRR')
ax.grid(axis='y', alpha=0.2)
plt.tight_layout()
```
