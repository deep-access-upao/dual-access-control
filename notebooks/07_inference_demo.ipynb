```python
from pathlib import Path
import sys

PROJECT_ROOT = Path.cwd().resolve()
if PROJECT_ROOT.name == 'notebooks':
    PROJECT_ROOT = PROJECT_ROOT.parent
if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

from src.inference import FaceVerifier, decide_access, load_model_config

CONFIG_PATH = PROJECT_ROOT / 'config/model_config.json'
config = load_model_config(CONFIG_PATH)
{
    'model_name': config.model_name,
    'threshold': config.threshold,
    'score_rule': config.score_rule,
    'input_size': config.input_size,
}
```

## Rutas editables

Usa únicamente imágenes autorizadas del dataset/support set. La celda siguiente está desactivada hasta completar rutas JPEG válidas.

```python
REFERENCE_IMAGE = PROJECT_ROOT / 'data/support_set/usuario/frontal.jpg'
CAPTURE_IMAGE = PROJECT_ROOT / 'data/processed/ruta_autorizada/captura.jpg'
RUN_INFERENCE = False

if RUN_INFERENCE:
    verifier = FaceVerifier.from_config(CONFIG_PATH)
    result = verifier.verify_pair(REFERENCE_IMAGE, CAPTURE_IMAGE)
    display(result.to_dict())
else:
    print('Edita las rutas y cambia RUN_INFERENCE=True para ejecutar.')
```
